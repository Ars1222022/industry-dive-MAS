# Vad är ett Multi-Agent System (MAS)? 

```python
# ========================================
# CELL 1: VAD ÄR ETT MULTI-AGENT SYSTEM (MAS)?
# ========================================

print("=" * 60)
print("🤖 MULTI-AGENT SYSTEM (MAS) – KODEXEMPEL")
print("=" * 60)

# ----------------------------------------
# UTAN MAS (en funktion gör allt) – DÅLIGT
# ----------------------------------------
print("\n❌ UTAN MAS (en funktion gör allt):")
print("-" * 40)

def do_everything(role):
    # En funktion försöker göra ALLT
    print("   1. Genererar sökord...")
    keywords = ["AI engineer", "ML engineer", "data scientist"]
    
    print("   2. Hämtar jobb...")
    jobs = [f"Jobb som {kw}" for kw in keywords[:2]]
    
    print("   3. Analyserar tech-stack...")
    tech_stack = {"Python": 80, "AWS": 60}
    
    print("   4. Rekommenderar certifieringar...")
    certs = ["AWS Certified ML", "Databricks Associate"]
    
    return certs

result = do_everything("AI developer")
print(f"\n   Resultat: {result}")
print("   ⚠️ PROBLEM: Svårt att ändra, testa eller återanvända delar!")

# ----------------------------------------
# MED MAS (flera funktioner samarbetar) – BRA
# ----------------------------------------
print("\n✅ MED MAS (flera funktioner samarbetar):")
print("-" * 40)

# Agent 1: Bara sökord
def agent_generate_keywords(role):
    print("   🤖 Agent 1: Genererar sökord")
    return ["AI engineer", "ML engineer", "data scientist"]

# Agent 2: Bara hämta jobb
def agent_fetch_jobs(keyword):
    print("   🤖 Agent 2: Hämtar jobb")
    return [f"Jobb: {keyword} hos Företag X", f"Jobb: {keyword} hos Företag Y"]

# Agent 3: Bara analysera stack
def agent_analyze_stack(jobs):
    print("   🤖 Agent 3: Analyserar tech-stack")
    return {"Python": 80, "AWS": 60, "SQL": 50}

# Agent 4: Bara certifieringar
def agent_recommend_certs(stack):
    print("   🤖 Agent 4: Rekommenderar certifieringar")
    return ["AWS Certified ML", "Databricks Associate", "TensorFlow Developer"]

# Alla agenter jobbar tillsammans
keywords = agent_generate_keywords("AI developer")
all_jobs = []
for kw in keywords[:2]:  # Ta första 2 för demo
    all_jobs.extend(agent_fetch_jobs(kw))

stack = agent_analyze_stack(all_jobs)
certs = agent_recommend_certs(stack)

print(f"\n   🎯 Resultat: {certs}")
print("   ✅ FÖRDELAR: Lätt att ändra, testa och återanvända varje agent!")

# ----------------------------------------
# JÄMFÖRELSE – SIDVIDES
# ----------------------------------------
print("\n📊 JÄMFÖRELSE:")
print("-" * 40)
print("""
   Utan MAS: 1 funktion → 20 rader kod, svår att ändra
   Med MAS:  4 funktioner → 4 x 5 rader = 20 rader, LÄTT att ändra!
   
   💡 Vill du byta API för jobb-hämtning? Ändra bara Agent 2!
   💡 Vill du lägga till en ny agent? Bara lägg till!
""")

print("\n" + "=" * 60)
print("✅ KLART! Nästa cell: Installera bibliotek")
print("=" * 60)
```

```python
# ========================================
# CELL 2: AGENT 2 – JOBB-HÄMTARE (MED MENY FÖR PERIOD OCH REGION)
# ========================================

import requests
import json
import time
from datetime import datetime, timedelta

print("=" * 60)
print("🤖 AGENT 2: JOBB-HÄMTARE")
print("=" * 60)

# ----------------------------------------
# HÄMTA SÖKORD FRÅN AGENT 1
# ----------------------------------------
result = spark.sql("SELECT * FROM agent1_keywords").collect()
if len(result) == 0:
    print("\n❌ Inga sökord hittades! Kör CELL 1 först.")
else:
    keywords = json.loads(result[0]['keywords'])
    print(f"\n📋 Sökord från Agent 1: {keywords}")

    # ----------------------------------------
    # MENY 1: VÄLJ TIDSPERIOD
    # ----------------------------------------
    print("\n" + "=" * 60)
    print("📅 VÄLJ TIDSPERIOD")
    print("=" * 60)
    print("""
    ┌─────────────────────────────────────────────────────────────────┐
    │  Val  │  Period                                               │
    ├───────┼───────────────────────────────────────────────────────┤
    │  1    │  Senaste 7 dagarna                                    │
    │  2    │  7-14 dagar sedan                                      │
    │  3    │  14-21 dagar sedan                                     │
    │  4    │  21-28 dagar sedan                                     │
    │  5    │  ALLA perioder (sparas separat för trendanalys)        │
    │  6    │  Anpassad period (ange egna dagar)                     │
    └─────────────────────────────────────────────────────────────────┘
    """)
    
    period_choice = input("   👉 Välj (1-6): ").strip()
    
    # Beräkna datumintervall
    today = datetime.now()
    
    if period_choice == '1':
        period_name = "senaste_7_dagar"
        start_date = (today - timedelta(days=7)).strftime("%Y-%m-%dT00:00:00")
        end_date = today.strftime("%Y-%m-%dT23:59:59")
        print(f"\n   ✅ Hämtar jobb från senaste 7 dagarna")
        
    elif period_choice == '2':
        period_name = "dag_7_14"
        start_date = (today - timedelta(days=14)).strftime("%Y-%m-%dT00:00:00")
        end_date = (today - timedelta(days=7)).strftime("%Y-%m-%dT23:59:59")
        print(f"\n   ✅ Hämtar jobb från 7-14 dagar sedan")
        
    elif period_choice == '3':
        period_name = "dag_14_21"
        start_date = (today - timedelta(days=21)).strftime("%Y-%m-%dT00:00:00")
        end_date = (today - timedelta(days=14)).strftime("%Y-%m-%dT23:59:59")
        print(f"\n   ✅ Hämtar jobb från 14-21 dagar sedan")
        
    elif period_choice == '4':
        period_name = "dag_21_28"
        start_date = (today - timedelta(days=28)).strftime("%Y-%m-%dT00:00:00")
        end_date = (today - timedelta(days=21)).strftime("%Y-%m-%dT23:59:59")
        print(f"\n   ✅ Hämtar jobb från 21-28 dagar sedan")
        
    elif period_choice == '5':
        period_name = "alla_perioder"
        print("\n   ✅ Hämtar jobb från ALLA perioder (sparas separat)")
        
    elif period_choice == '6':
        print("\n   📅 Anpassad period:")
        days_back = int(input("      Hur många dagar tillbaka? "))
        start_date = (today - timedelta(days=days_back)).strftime("%Y-%m-%dT00:00:00")
        end_date = today.strftime("%Y-%m-%dT23:59:59")
        period_name = f"anpassad_{days_back}_dagar"
        print(f"\n   ✅ Hämtar jobb från senaste {days_back} dagarna")
        
    else:
        print("\n   ⚠️ Ogiltigt val! Använder senaste 7 dagarna.")
        period_name = "senaste_7_dagar"
        start_date = (today - timedelta(days=7)).strftime("%Y-%m-%dT00:00:00")
        end_date = today.strftime("%Y-%m-%dT23:59:59")

    # ----------------------------------------
    # MENY 2: VÄLJ REGION
    # ----------------------------------------
    print("\n" + "=" * 60)
    print("🗺️ VÄLJ REGION")
    print("=" * 60)
    print("""
    ┌─────────────────────────────────────────────────────────────────┐
    │  Val  │  Region-kod  │  Område                                 │
    ├───────┼──────────────┼─────────────────────────────────────────┤
    │  1    │  (ingen)     │  Hela Sverige                           │
    │  2    │  01          │  Stockholms län                         │
    │  3    │  03          │  Uppsala län                            │
    │  4    │  04          │  Södermanlands län                      │
    │  5    │  05          │  Östergötlands län                      │
    │  6    │  06          │  Jönköpings län                         │
    │  7    │  07          │  Kronobergs län                         │
    │  8    │  08          │  Kalmar län                             │
    │  9    │  09          │  Gotlands län                           │
    │  10   │  10          │  Blekinge län                           │
    │  11   │  12          │  Skåne län                              │
    │  12   │  13          │  Hallands län                           │
    │  13   │  14          │  Västra Götalands län                   │
    │  14   │  17          │  Värmlands län                          │
    │  15   │  18          │  Örebro län                             │
    │  16   │  19          │  Västmanlands län                       │
    │  17   │  20          │  Dalarnas län                           │
    │  18   │  21          │  Gävleborgs län                         │
    │  19   │  22          │  Västernorrlands län                    │
    │  20   │  23          │  Jämtlands län                          │
    │  21   │  24          │  Västerbottens län                      │
    │  22   │  25          │  Norrbottens län                        │
    └─────────────────────────────────────────────────────────────────┘
    """)
    
    region_choice = input("   👉 Välj region (1-22): ").strip()
    
    region_map = {
        '1': '', '2': '01', '3': '03', '4': '04', '5': '05', '6': '06',
        '7': '07', '8': '08', '9': '09', '10': '10', '11': '12', '12': '13',
        '13': '14', '14': '17', '15': '18', '16': '19', '17': '20', '18': '21',
        '19': '22', '20': '23', '21': '24', '22': '25'
    }
    
    region_codes = {
        '1': 'Hela Sverige', '2': 'Stockholms län', '3': 'Uppsala län',
        '4': 'Södermanlands län', '5': 'Östergötlands län', '6': 'Jönköpings län',
        '7': 'Kronobergs län', '8': 'Kalmar län', '9': 'Gotlands län',
        '10': 'Blekinge län', '11': 'Skåne län', '12': 'Hallands län',
        '13': 'Västra Götalands län', '14': 'Värmlands län', '15': 'Örebro län',
        '16': 'Västmanlands län', '17': 'Dalarnas län', '18': 'Gävleborgs län',
        '19': 'Västernorrlands län', '20': 'Jämtlands län', '21': 'Västerbottens län',
        '22': 'Norrbottens län'
    }
    
    region_code = region_map.get(region_choice, '')
    region_name = region_codes.get(region_choice, 'Hela Sverige')
    
    print(f"\n   ✅ Region: {region_name}")
    
    if region_code:
        print(f"   📍 Region-kod: {region_code}")

    # ----------------------------------------
    # HÄMTA JOBB FÖR VARJE SÖKORD
    # ----------------------------------------
    print("\n🔍 Hämtar jobb...")
    print("-" * 40)
    
    all_jobs = []
    job_details = []
    
    for keyword in keywords:
        # Bygg URL med region om vald
        if region_code:
            url = f"https://jobsearch.api.jobtechdev.se/search?q={keyword}&limit=20&region={region_code}"
        else:
            url = f"https://jobsearch.api.jobtechdev.se/search?q={keyword}&limit=20"
        
        # Lägg till tidsintervall om valt (förutom "alla perioder")
        if period_choice != '5' and period_choice != '':
            url += f"&published-after={start_date}&published-before={end_date}"
        
        try:
            response = requests.get(url)
            if response.status_code == 200:
                data = response.json()
                hits = data.get('hits', [])
                total = data.get('total', {}).get('value', 0)
                
                print(f"\n   📌 '{keyword}': {total} jobb hittades")
                
                for job in hits[:10]:
                    title = job.get('headline', 'Okänd titel')
                    employer = job.get('employer', {}).get('name', 'Okänd arbetsgivare')
                    description = job.get('description', {}).get('text', '')[:500]
                    pub_date = job.get('publication_date', '')
                    
                    print(f"      - {title} hos {employer} ({pub_date[:10] if pub_date else 'datum okänt'})")
                    
                    job_details.append({
                        'keyword': keyword,
                        'title': title,
                        'employer': employer,
                        'description': description,
                        'publication_date': pub_date
                    })
                    all_jobs.append(job)
            else:
                print(f"\n   ⚠️ Fel vid '{keyword}': {response.status_code}")
                
        except Exception as e:
            print(f"\n   ❌ Fel vid '{keyword}': {e}")
        
        time.sleep(0.2)
    
    print(f"\n✅ TOTALT: {len(all_jobs)} jobb hittades i {region_name}")
    
    # ----------------------------------------
    # SPARA RESULTAT FÖR AGENT 3 (med metadata om period och region)
    # ----------------------------------------
    print("\n💾 Sparar resultat för Agent 3...")
    print("-" * 40)
    
    # Spara både sökord, jobb och metadata
    metadata = {
        'period': period_name,
        'region_code': region_code,
        'region_name': region_name,
        'search_date': today.strftime("%Y-%m-%d %H:%M:%S"),
        'total_jobs': len(job_details)
    }
    
    keywords_json = json.dumps(keywords)
    jobs_json = json.dumps(job_details)
    metadata_json = json.dumps(metadata)
    
    spark.createDataFrame([{"keywords": keywords_json, "jobs": jobs_json, "metadata": metadata_json}]) \
          .createOrReplaceTempView("agent2_results")
    
    print(f"   ✅ Sökord och jobb sparade!")
    print(f"   📋 Sökord: {len(keywords)} st")
    print(f"   📊 Antal jobb: {len(job_details)}")
    print(f"   📅 Period: {period_name}")
    print(f"   🗺️ Region: {region_name}")
    
    # ----------------------------------------
    # SPARA I SEPARATA FILER FÖR OLIKA PERIODER (för trendanalys)
    # ----------------------------------------
    if period_choice == '5':
        print("\n📁 Sparar jobb per period för trendanalys...")
        
        periods = [
            ('senaste_7_dagar', 7, 0),
            ('dag_7_14', 14, 7),
            ('dag_14_21', 21, 14),
            ('dag_21_28', 28, 21)
        ]
        
        for period, days_back, days_offset in periods:
            start = (today - timedelta(days=days_back)).strftime("%Y-%m-%dT00:00:00")
            end = (today - timedelta(days=days_offset)).strftime("%Y-%m-%dT23:59:59") if days_offset > 0 else today.strftime("%Y-%m-%dT23:59:59")
            
            period_jobs = []
            for keyword in keywords:
                if region_code:
                    url = f"https://jobsearch.api.jobtechdev.se/search?q={keyword}&limit=10&region={region_code}&published-after={start}&published-before={end}"
                else:
                    url = f"https://jobsearch.api.jobtechdev.se/search?q={keyword}&limit=10&published-after={start}&published-before={end}"
                
                try:
                    response = requests.get(url)
                    if response.status_code == 200:
                        data = response.json()
                        for job in data.get('hits', []):
                            period_jobs.append(job)
                except:
                    pass
                time.sleep(0.1)
            
            print(f"   📊 {period}: {len(period_jobs)} jobb")
            
            # Spara periodens resultat separat
            period_metadata = {
                'period': period,
                'region_name': region_name,
                'total_jobs': len(period_jobs)
            }
            
            spark.createDataFrame([{"keywords": keywords_json, "jobs": json.dumps(period_jobs), "metadata": json.dumps(period_metadata)}]) \
                  .createOrReplaceTempView(f"agent2_results_{period}")
    
    print("\n" + "=" * 60)
    print("✅ AGENT 2 KLAR! Kör nu CELL 3 (Agent 3)")
    print("=" * 60)
```

```python
# ========================================
# CELL 3: AGENT 3 – STACK-ANALYSATOR (MED GRAFISKA VISUALISERINGAR)
# ========================================

import json
import requests
import getpass
from collections import Counter
from datetime import datetime, timedelta
import pandas as pd
import os

# Importera visualiseringsbibliotek
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# För interaktiva grafer (valfritt)
try:
    import plotly.express as px
    import plotly.graph_objects as go
    PLOTLY_AVAILABLE = True
except ImportError:
    PLOTLY_AVAILABLE = False

# Sätt svenskt typsnitt och stil
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Helvetica']
sns.set_style("whitegrid")
sns.set_palette("husl")

print("=" * 60)
print("🤖 AGENT 3: STACK-ANALYSATOR (med grafer)")
print("=" * 60)

# ----------------------------------------
# Fråga efter Groq API-nyckel
# ----------------------------------------
print("\n🔑 Ange din Groq API-nyckel:")
GROQ_KEY = getpass.getpass("   ➤ ")

# ----------------------------------------
# Hämta jobb från Agent 2
# ----------------------------------------
result = spark.sql("SELECT * FROM agent2_results").collect()
if len(result) == 0:
    print("❌ Kör CELL 1 och CELL 2 först!")
else:
    keywords = json.loads(result[0]['keywords'])
    jobs = json.loads(result[0]['jobs'])
    metadata = json.loads(result[0]['metadata'])
    
    print(f"\n📊 ANALYS-INFORMATION:")
    print(f"   ✅ Jobb att analysera: {len(jobs)}")
    print(f"   📅 Period: {metadata.get('period', 'okänd')}")
    print(f"   🗺️ Region: {metadata.get('region_name', 'okänd')}")

    # ----------------------------------------
    # FUNKTION: Hämta datumintervall
    # ----------------------------------------
    def get_date_ranges():
        today = datetime.now()
        return {
            'senaste_7_dagar': {
                'start': (today - timedelta(days=7)).strftime("%Y-%m-%dT00:00:00"),
                'end': today.strftime("%Y-%m-%dT23:59:59"),
                'name': 'Vecka 1 (senaste 7)',
                'order': 1
            },
            'dag_7_14': {
                'start': (today - timedelta(days=14)).strftime("%Y-%m-%dT00:00:00"),
                'end': (today - timedelta(days=7)).strftime("%Y-%m-%dT23:59:59"),
                'name': 'Vecka 2 (7-14 dagar)',
                'order': 2
            },
            'dag_14_21': {
                'start': (today - timedelta(days=21)).strftime("%Y-%m-%dT00:00:00"),
                'end': (today - timedelta(days=14)).strftime("%Y-%m-%dT23:59:59"),
                'name': 'Vecka 3 (14-21 dagar)',
                'order': 3
            },
            'dag_21_28': {
                'start': (today - timedelta(days=28)).strftime("%Y-%m-%dT00:00:00"),
                'end': (today - timedelta(days=21)).strftime("%Y-%m-%dT23:59:59"),
                'name': 'Vecka 4 (21-28 dagar)',
                'order': 4
            }
        }

    # ----------------------------------------
    # FUNKTION: Hämta jobb från API för en period
    # ----------------------------------------
    def fetch_jobs_for_period(period_name, start_date, end_date, keywords_list, region_code=''):
        period_jobs = []
        
        for keyword in keywords_list:
            if region_code:
                url = f"https://jobsearch.api.jobtechdev.se/search?q={keyword}&limit=10&region={region_code}&published-after={start_date}&published-before={end_date}"
            else:
                url = f"https://jobsearch.api.jobtechdev.se/search?q={keyword}&limit=10&published-after={start_date}&published-before={end_date}"
            
            try:
                response = requests.get(url)
                if response.status_code == 200:
                    data = response.json()
                    for job in data.get('hits', []):
                        period_jobs.append({
                            'keyword': keyword,
                            'title': job.get('headline', 'Okänd titel'),
                            'employer': job.get('employer', {}).get('name', 'Okänd arbetsgivare'),
                            'description': job.get('description', {}).get('text', '')[:500],
                            'publication_date': job.get('publication_date', '')
                        })
            except Exception as e:
                pass
            
            import time
            time.sleep(0.1)
        
        return period_jobs

    # ----------------------------------------
    # FUNKTION: Keyword-baserad analys (fallback)
    # ----------------------------------------
    def keyword_based_analysis(job_list):
        tech_keywords = [
            "Python", "Java", "JavaScript", "TypeScript", "SQL", "C#", "C++",
            "AWS", "Azure", "GCP", "Docker", "Kubernetes", "Git", "Terraform",
            "TensorFlow", "PyTorch", "scikit-learn", "Pandas", "NumPy",
            "React", "Angular", "Vue", "Node.js", "Django", "Flask", "Spring",
            "Spark", "Databricks", "Snowflake", "LangChain", "OpenAI", "MLflow",
            "Power BI", "Tableau", "BigQuery", "Airflow", "Hadoop", "Kafka",
            "Machine Learning", "AI", "Deep Learning", "NLP", "Computer Vision"
        ]
        
        tech_counter = Counter()
        for job in job_list:
            title = job.get('title', '')
            description = job.get('description', '')[:1000]
            text = f"{title} {description}".lower()
            
            for tech in tech_keywords:
                if tech.lower() in text:
                    tech_counter[tech] += 1
        
        return tech_counter

    # ----------------------------------------
    # FUNKTION: Skapa trendgraf med matplotlib
    # ----------------------------------------
    def create_trend_chart(trend_data, top_n=8):
        """Skapar en trendgraf över tid med matplotlib"""
        
        if not trend_data:
            print("   Ingen data för graf")
            return None
        
        # Förbered data
        periods = []
        period_names = []
        for key, data in sorted(trend_data.items(), key=lambda x: x[1].get('order', 0)):
            periods.append(key)
            period_names.append(data.get('name', key))
        
        # Hitta topp-teknologier
        all_techs = Counter()
        for period_key in periods:
            all_techs.update(trend_data[period_key].get('counter', Counter()))
        
        top_techs = [tech for tech, _ in all_techs.most_common(top_n)]
        
        # Skapa DataFrame för plotting
        df_data = []
        for period_key in periods:
            period_counter = trend_data[period_key].get('counter', Counter())
            period_name = trend_data[period_key].get('name', period_key)
            for tech in top_techs:
                df_data.append({
                    'Period': period_name,
                    'Technology': tech,
                    'Count': period_counter.get(tech, 0)
                })
        
        df = pd.DataFrame(df_data)
        
        # Skapa grafen
        fig, ax = plt.subplots(figsize=(14, 8))
        
        # Använd seaborn för finare utseende
        sns.lineplot(data=df, x='Period', y='Count', hue='Technology', 
                    marker='o', markersize=8, linewidth=2.5, ax=ax)
        
        # Anpassa utseende
        ax.set_title(f'📈 Trendanalys: Efterfrågan på teknologier över tid\n({metadata.get("region_name", "Sverige")})', 
                    fontsize=16, fontweight='bold', pad=20)
        ax.set_xlabel('Tidsperiod', fontsize=12, fontweight='bold')
        ax.set_ylabel('Antal jobbannonser', fontsize=12, fontweight='bold')
        ax.legend(title='Teknologi', bbox_to_anchor=(1.05, 1), loc='upper left')
        ax.grid(True, alpha=0.3)
        
        # Rotera x-axeln om det behövs
        plt.xticks(rotation=15, ha='right')
        
        # Justera layout
        plt.tight_layout()
        
        # Visa i notebook
        display(fig)
        
        return fig

    # ----------------------------------------
    # FUNKTION: Skapa stapeldiagram för topp-teknologier
    # ----------------------------------------
    def create_bar_chart(tech_counter, title="Populäraste teknologier"):
        """Skapar ett horisontellt stapeldiagram"""
        
        if not tech_counter:
            return None
        
        # Ta top 15
        top_items = tech_counter.most_common(15)
        technologies = [item[0] for item in top_items]
        counts = [item[1] for item in top_items]
        
        # Skapa figur
        fig, ax = plt.subplots(figsize=(12, 8))
        
        # Horisontellt stapeldiagram
        bars = ax.barh(technologies, counts, color=sns.color_palette("viridis", len(technologies)))
        
        # Lägg till värden på staplarna
        for i, (bar, count) in enumerate(zip(bars, counts)):
            ax.text(count + 0.5, bar.get_y() + bar.get_height()/2, 
                   f'{count}', va='center', fontsize=10, fontweight='bold')
        
        ax.set_xlabel('Antal förekomster', fontsize=12, fontweight='bold')
        ax.set_title(title, fontsize=14, fontweight='bold', pad=15)
        ax.invert_yaxis()  # Så att högsta är överst
        ax.grid(True, alpha=0.3, axis='x')
        
        plt.tight_layout()
        display(fig)
        
        return fig

    # ----------------------------------------
    # FUNKTION: Skapa heatmap för trenddata
    # ----------------------------------------
    def create_heatmap(trend_data, top_n=10):
        """Skapar en heatmap över teknologier över tid"""
        
        if not trend_data:
            return None
        
        # Förbered data
        periods = []
        period_names = []
        for key, data in sorted(trend_data.items(), key=lambda x: x[1].get('order', 0)):
            periods.append(key)
            period_names.append(data.get('name', key))
        
        # Hitta topp-teknologier
        all_techs = Counter()
        for period_key in periods:
            all_techs.update(trend_data[period_key].get('counter', Counter()))
        
        top_techs = [tech for tech, _ in all_techs.most_common(top_n)]
        
        # Skapa matris för heatmap
        heatmap_data = []
        for tech in top_techs:
            row = []
            for period_key in periods:
                period_counter = trend_data[period_key].get('counter', Counter())
                row.append(period_counter.get(tech, 0))
            heatmap_data.append(row)
        
        # Skapa heatmap
        fig, ax = plt.subplots(figsize=(12, 8))
        
        im = ax.imshow(heatmap_data, cmap='YlOrRd', aspect='auto')
        
        # Anpassa axlar
        ax.set_xticks(range(len(period_names)))
        ax.set_xticklabels(period_names, rotation=45, ha='right')
        ax.set_yticks(range(len(top_techs)))
        ax.set_yticklabels(top_techs)
        
        # Lägg till färgskala
        plt.colorbar(im, ax=ax, label='Antal jobbannonser')
        
        # Lägg till siffror i rutorna
        for i in range(len(top_techs)):
            for j in range(len(period_names)):
                text = ax.text(j, i, heatmap_data[i][j],
                             ha="center", va="center", color="black", fontsize=9)
        
        ax.set_title(f'📊 Heatmap: Teknologier över tid\n({metadata.get("region_name", "Sverige")})', 
                    fontsize=14, fontweight='bold', pad=20)
        
        plt.tight_layout()
        display(fig)
        
        return fig

    # ----------------------------------------
    # FUNKTION: Skapa interaktiv graf med Plotly (om tillgängligt)
    # ----------------------------------------
    def create_interactive_chart(trend_data, top_n=8):
        """Skapar en interaktiv graf med Plotly"""
        
        if not PLOTLY_AVAILABLE:
            print("   📌 Plotly är inte installerat. Använder statiska grafer istället.")
            return None
        
        if not trend_data:
            return None
        
        # Förbered data
        periods = []
        period_names = []
        for key, data in sorted(trend_data.items(), key=lambda x: x[1].get('order', 0)):
            periods.append(key)
            period_names.append(data.get('name', key))
        
        # Hitta topp-teknologier
        all_techs = Counter()
        for period_key in periods:
            all_techs.update(trend_data[period_key].get('counter', Counter()))
        
        top_techs = [tech for tech, _ in all_techs.most_common(top_n)]
        
        # Skapa DataFrame
        df_data = []
        for period_key in periods:
            period_counter = trend_data[period_key].get('counter', Counter())
            period_name = trend_data[period_key].get('name', period_key)
            for tech in top_techs:
                df_data.append({
                    'Period': period_name,
                    'Technology': tech,
                    'Count': period_counter.get(tech, 0)
                })
        
        df = pd.DataFrame(df_data)
        
        # Skapa interaktiv graf
        fig = px.line(df, x='Period', y='Count', color='Technology',
                     title=f'Trendanalys: Efterfrågan på teknologier över tid<br><sub>{metadata.get("region_name", "Sverige")}</sub>',
                     markers=True, line_shape='linear')
        
        fig.update_layout(
            xaxis_title='Tidsperiod',
            yaxis_title='Antal jobbannonser',
            hovermode='x unified',
            legend_title='Teknologi',
            width=1000,
            height=600
        )
        
        display(fig)
        return fig

    # ----------------------------------------
    # HUVUDANALYS
    # ----------------------------------------
    
    if metadata.get('period') == 'alla_perioder':
        print("\n" + "=" * 60)
        print("📊 TRENDANALYS – ALLA PERIODER")
        print("=" * 60)
        
        region_code = metadata.get('region_code', '')
        date_ranges = get_date_ranges()
        trend_data = {}
        
        for period_key, period_info in date_ranges.items():
            print(f"\n📅 Hämtar period: {period_info['name']}")
            period_jobs = fetch_jobs_for_period(
                period_key,
                period_info['start'],
                period_info['end'],
                keywords,
                region_code
            )
            
            if len(period_jobs) > 0:
                print(f"   🔧 Analyserar {len(period_jobs)} jobb...")
                period_counter = keyword_based_analysis(period_jobs)  # Använd keyword-baserad för hastighet
                trend_data[period_key] = {
                    'counter': period_counter,
                    'name': period_info['name'],
                    'order': period_info['order']
                }
            else:
                trend_data[period_key] = {
                    'counter': Counter(),
                    'name': period_info['name'],
                    'order': period_info['order']
                }
        
        # Sammanställ total statistik
        total_counter = Counter()
        for period_key in trend_data:
            total_counter.update(trend_data[period_key]['counter'])
        
        # ========================================
        # VISA GRAFISKA VISUALISERINGAR
        # ========================================
        
        print("\n" + "=" * 60)
        print("📈 GENERERAR VISUALISERINGAR")
        print("=" * 60)
        
        # 1. Trendgraf över tid
        print("\n📊 1. Trendgraf (linjediagram)")
        print("-" * 40)
        create_trend_chart(trend_data, top_n=8)
        
        # 2. Stapeldiagram för totalen
        print("\n📊 2. Total tech-stack (stapeldiagram)")
        print("-" * 40)
        create_bar_chart(total_counter, f"Populäraste teknologier i {metadata.get('region_name', 'Sverige')}")
        
        # 3. Heatmap för trender
        print("\n📊 3. Heatmap (teknologier över tid)")
        print("-" * 40)
        create_heatmap(trend_data, top_n=10)
        
        # 4. Interaktiv graf (om Plotly finns)
        if PLOTLY_AVAILABLE:
            print("\n📊 4. Interaktiv graf (Plotly)")
            print("-" * 40)
            create_interactive_chart(trend_data, top_n=8)
        
        # Visa tabell i terminalen
        print("\n📊 TOTAL TECH-STACK (alla perioder)")
        print("-" * 60)
        for tech, count in total_counter.most_common(15):
            pct = round(count / sum(total_counter.values()) * 100, 1) if sum(total_counter.values()) > 0 else 0
            bar = "█" * int(pct / 2)
            print(f"   {tech:<25} {count:3} gånger ({pct:5.1f}%) {bar}")
        
        # Fråga om export
        print("\n" + "-" * 60)
        export_choice = input("\n💾 Vill du exportera data? (j/n): ").strip().lower()
        if export_choice == 'j':
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            region_name = metadata.get('region_name', 'sverige').replace(' ', '_')
            
            # Spara trenddata som CSV
            trend_df_data = []
            for period_key, period_data in trend_data.items():
                for tech, count in period_data['counter'].most_common(20):
                    trend_df_data.append({
                        'Period': period_data['name'],
                        'Technology': tech,
                        'Count': count
                    })
            
            trend_df = pd.DataFrame(trend_df_data)
            csv_filename = f"trend_analysis_{region_name}_{timestamp}.csv"
            trend_df.to_csv(csv_filename, index=False, encoding='utf-8-sig')
            print(f"\n   ✅ Data sparad: {csv_filename}")
            
            # Spara grafer som bilder
            fig1 = create_trend_chart(trend_data, top_n=8)
            if fig1:
                fig1.savefig(f"trend_chart_{region_name}_{timestamp}.png", dpi=150, bbox_inches='tight')
                print(f"   ✅ Graf sparad: trend_chart_{region_name}_{timestamp}.png")
    
    else:
        # Vanlig analys för en period
        print("\n🔧 Analyserar tech-stack...")
        tech_counter = keyword_based_analysis(jobs)
        
        # Visa stapeldiagram
        create_bar_chart(tech_counter, f"Populäraste teknologier ({metadata.get('region_name', '')}, {metadata.get('period', '')})")
        
        # Visa tabell
        display_results(tech_counter, f"POPULÄRASTE TECH-STACKEN")
        
        # Fråga om export
        export_choice = input("\n💾 Vill du exportera resultaten? (j/n): ").strip().lower()
        if export_choice == 'j':
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            region_name = metadata.get('region_name', 'sverige').replace(' ', '_')
            
            df_data = []
            for tech, count in tech_counter.most_common():
                df_data.append({'Technology': tech, 'Count': count})
            
            df = pd.DataFrame(df_data)
            csv_filename = f"tech_stack_{region_name}_{timestamp}.csv"
            df.to_csv(csv_filename, index=False, encoding='utf-8-sig')
            print(f"\n   ✅ Data sparad: {csv_filename}")
    
    print("\n" + "=" * 60)
    print("✅ AGENT 3 KLAR!")
    print("=" * 60)
```

Agent 4. CERTIFIERINGSRECOMMENDATIONER (Tavily + Groq)

```python
# ========================================
# CELL 4: AGENT 4 – CERTIFIERINGAR (ALLA 10 + FIN FORMAT + POWER BI READY)
# ========================================

import json
import requests
import getpass
from collections import Counter
import time
import pandas as pd
from datetime import datetime

print("=" * 60)
print("🤖 AGENT 4: CERTIFIERINGSRECOMMENDATIONER")
print("=" * 60)

# ----------------------------------------
# FRÅGA EFTER API NYCKLAR
# ----------------------------------------
print("\n🔑 Ange din Groq API-nyckel:")
GROQ_KEY = getpass.getpass("   ➤ Groq: ")

print("\n🔑 Ange din Tavily API-nyckel:")
TAVILY_KEY = getpass.getpass("   ➤ Tavily: ")

# ----------------------------------------
# HÄMTA RESULTAT FRÅN AGENT 3
# ----------------------------------------
print("\n📥 Hämtar tech-stack från Agent 3...")

result = spark.sql("SELECT * FROM agent3_results").collect()
if len(result) == 0:
    print("❌ Ingen data från Agent 3! Kör Agent 3 först.")
else:
    agent3_data = json.loads(result[0]['agent3_data'])
    tech_stack = agent3_data.get('tech_stack', {})
    metadata = agent3_data.get('metadata', {})
    
    print(f"   ✅ Hittade {len(tech_stack)} teknologier att analysera")
    
    # Sortera och visa alla
    tech_items = sorted(tech_stack.items(), key=lambda x: x[1], reverse=True)
    
    print("\n📊 ALLA TEKNOLOGIER FRÅN AGENT 3:")
    print("-" * 50)
    for i, (tech, count) in enumerate(tech_items, 1):
        bar = "█" * min(count, 20)
        print(f"   {i:2}. {tech:<20} ({count:2} förekomster) {bar}")
    
    # ----------------------------------------
    # ANALYSERA ALLA 10 (eller färre om mindre finns)
    # ----------------------------------------
    top_n = min(10, len(tech_stack))
    print(f"\n🔍 Analyserar de {top_n} populäraste teknologierna...")
    tech_items_analysis = tech_items[:top_n]
    
    # ----------------------------------------
    # FUNKTIONER FÖR SÖKNING OCH ANALYS
    # ----------------------------------------
    def search_certifications(technology, max_results=3):
        url = "https://api.tavily.com/search"
        headers = {"Authorization": f"Bearer {TAVILY_KEY}", "Content-Type": "application/json"}
        
        query_map = {
            "Git": f"{technology} certification training online 2025",
            "Python": f"Python certification PCAP PCEP course 2025",
            "Java": f"Oracle Java certification SE 2025",
            "C#": f"Microsoft C# certification .NET 2025",
            "Databricks": f"Databricks certification data engineer 2025",
            "AWS": f"AWS certification cloud practitioner 2025",
            "Docker": f"Docker certification DCA 2025",
            "Kubernetes": f"Kubernetes certification CKA 2025",
            "Spark": f"Apache Spark certification Databricks 2025",
            "TensorFlow": f"TensorFlow certification Google 2025"
        }
        query = query_map.get(technology, f"best {technology} certification course online 2025")
        
        data = {"query": query, "search_depth": "basic", "max_results": max_results}
        
        try:
            response = requests.post(url, headers=headers, json=data)
            if response.status_code == 200:
                return response.json()
        except:
            pass
        return {"results": []}
    
    def analyze_certifications(technology, search_results):
        url = "https://api.groq.com/openai/v1/chat/completions"
        headers = {"Authorization": f"Bearer {GROQ_KEY}", "Content-Type": "application/json"}
        
        results_text = ""
        for result in search_results.get("results", [])[:3]:
            results_text += f"\nTitel: {result.get('title', '')}\nKälla: {result.get('url', '')}\nBeskrivning: {result.get('content', '')[:250]}\n"
        
        prompt = f"""
        Du är en karriärcoach. Ge EN rekommendation för {technology}.
        
        Sökresultat: {results_text}
        
        Svara EXAKT i detta JSON-format (ingen annan text):
        {{
            "technology": "{technology}",
            "certification": "Namn på certifiering",
            "provider": "Utgivare",
            "price": "Pris i SEK eller 'Gratis'",
            "hours": "Antal timmar",
            "url": "Länk till kurs/certifiering",
            "why": "Kort förklaring (max 20 ord)",
            "alternatives": "Alternativ 1, Alternativ 2"
        }}
        """
        
        data = {
            "model": "llama-3.3-70b-versatile",
            "messages": [{"role": "user", "content": prompt}],
            "temperature": 0.3,
            "response_format": {"type": "json_object"}
        }
        
        try:
            response = requests.post(url, headers=headers, json=data)
            if response.status_code == 200:
                return json.loads(response.json()['choices'][0]['message']['content'])
        except:
            pass
        return {"technology": technology, "certification": "Kunde inte hitta", "provider": "-", "price": "-", "hours": "-", "url": "-", "why": "-", "alternatives": "-"}
    
    # ----------------------------------------
    # HÄMTA CERTIFIERINGAR
    # ----------------------------------------
    certifications = []
    
    for i, (tech, count) in enumerate(tech_items_analysis, 1):
        print(f"\n   [{i}/{top_n}] {tech}...", end=" ", flush=True)
        search_results = search_certifications(tech)
        if search_results.get("results"):
            cert_data = analyze_certifications(tech, search_results)
            cert_data["frequency"] = count
            certifications.append(cert_data)
            print("✅")
        else:
            certifications.append({
                "technology": tech, "frequency": count,
                "certification": "Ingen data", "provider": "-", 
                "price": "-", "hours": "-", "url": "-", "why": "-", "alternatives": "-"
            })
            print("⚠️")
        time.sleep(0.5)
    
    # ----------------------------------------
    # VISA I FINT FORMAT (TABELL)
    # ----------------------------------------
    print("\n" + "=" * 80)
    print("📜 CERTIFIERINGSRECOMMENDATIONER")
    print("=" * 80)
    
    # Skapa fin tabell
    print(f"\n{'#'*80}")
    for cert in certifications:
        print(f"\n🔷 {cert['technology'].upper()} (förekommer i {cert['frequency']} jobb)")
        print(f"   {'─'*70}")
        print(f"   📜 Certifiering:  {cert['certification']}")
        print(f"   🏢 Utgivare:      {cert['provider']}")
        print(f"   💰 Pris:          {cert['price']}")
        print(f"   ⏱️ Tidsåtgång:    {cert['hours']} timmar")
        print(f"   🔗 Länk:          {cert['url']}")
        print(f"   📝 Varför:        {cert['why']}")
        print(f"   💡 Alternativ:    {cert['alternatives']}")
    
    # ----------------------------------------
    # SPARA FÖR POWER BI (JSON + CSV)
    # ----------------------------------------
    print("\n" + "=" * 80)
    print("💾 SPARAR FÖR POWER BI")
    print("=" * 80)
    
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    region_name = metadata.get('region_name', 'sverige').replace(' ', '_')
    
    # 1. JSON-format för automatisk laddning
    powerbi_data = {
        "metadata": {
            "export_date": timestamp,
            "region": metadata.get('region_name', 'Sverige'),
            "period": metadata.get('period', 'alla_perioder'),
            "total_technologies": len(tech_stack),
            "total_jobs_analyzed": metadata.get('jobs_count', 0)
        },
        "technologies": [
            {"technology": tech, "frequency": count} 
            for tech, count in tech_items
        ],
        "certifications": certifications
    }
    
    json_filename = f"powerbi_certifications_{region_name}_{timestamp}.json"
    with open(json_filename, 'w', encoding='utf-8') as f:
        json.dump(powerbi_data, f, ensure_ascii=False, indent=2)
    print(f"   ✅ JSON sparad: {json_filename}")
    
    # 2. CSV-format för Power BI
    df_cert = pd.DataFrame(certifications)
    csv_filename = f"powerbi_certifications_{region_name}_{timestamp}.csv"
    df_cert.to_csv(csv_filename, index=False, encoding='utf-8-sig')
    print(f"   ✅ CSV sparad: {csv_filename}")
    
    # 3. Tech-stack CSV
    df_tech = pd.DataFrame([{"Technology": tech, "Frequency": count} for tech, count in tech_items])
    tech_csv = f"powerbi_techstack_{region_name}_{timestamp}.csv"
    df_tech.to_csv(tech_csv, index=False, encoding='utf-8-sig')
    print(f"   ✅ Tech-stack CSV: {tech_csv}")
    
    # ----------------------------------------
    # SPARA I SPARK FÖR AGENT 5
    # ----------------------------------------
    spark.createDataFrame([{"agent4_data": json.dumps(powerbi_data)}]) \
          .createOrReplaceTempView("agent4_results")
    
    print("\n" + "=" * 80)
    print("📊 POWER BI INSTRUKTIONER")
    print("=" * 80)
    print(f"""
    Så här laddar du i Power BI:
    
    1. Öppna Power BI Desktop
    2. Klicka på "Hämta data" → "Text/CSV"
    3. Välj filen: {csv_filename}
    4. Klicka "Läs in"
    
    Eller för JSON:
    1. "Hämta data" → "JSON"
    2. Välj: {json_filename}
    
    Rekommenderade visualiseringar:
    - Stapeldiagram: Technology vs Frequency
    - Tabell: All certifieringsinfo
    - Kort: Totala jobb/teknologier
    """)
    
    print("\n" + "=" * 80)
    print("✅ AGENT 4 KLAR!")
    print("=" * 80)
```
